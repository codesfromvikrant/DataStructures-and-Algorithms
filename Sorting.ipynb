# Bubble Sort

```python
# swap elements with each other with iteration such that the bigger element shift in the last, repeat this iteration till the whole array get sorted.

arr = [22, 44, 99, 12, 54, 67, 10, 65, 32]

for turn in range(0, len(arr)-1):
  for j in range(0, len(arr)-turn-1):
    if(arr[j] > arr[j+1]):
      temp = arr[j]
      arr[j] = arr[j+1]
      arr[j+1] = temp

print(arr)

# Time Complexity is O(n^2)
```

# Selection Sort

```python
arr = [22, 44, 99, 12, 54, 67, 10, 65, 32]

for i in range(0, len(arr)):
  minPos = i
  for j in range(i+1,len(arr)):
    if(arr[j] < arr[minPos]):
        minPos = j
  temp = arr[i]
  arr[i] = arr[minPos]
  arr[minPos] = temp

print(arr)

# Time Complexity is O(n^2)
```

# Insertion Sort

```python
arr = [22, 44, 99, 12, 54, 67, 10, 65, 32]

sortLen = 1

for i in range(1, len(arr)):
  sortIndex = 0
  while(j<=sortLen):
    if(arr[j]>arr[i]):
      sortIndex = j
      break;
    else:
      j=j+1
  for k in range(i,sortIndex,-1):
    temp = arr[k]
    arr[k] = arr[k-1]
    arr[k-1] = temp
  sortLen = sortLen+1

print(arr)

# Time Complexity is O(n^2
```

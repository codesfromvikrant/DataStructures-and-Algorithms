# Bubble Sort

```python
arr = [22, 44, 99, 12, 54, 67, 10, 65, 32]
for i in range(0, len(arr)):
  for j in range(0, len(arr)-i-1):
    if(arr[j] > arr[j+1]):
      temp = arr[j]
      arr[j] = arr[j+1]
      arr[j+1] = temp

print(arr)

# Time Complexity is O(n^2)
```

```python
arr = [22, 44, 99, 12, 54, 67, 10, 65, 32]

for i in range(0, len(arr)):
  for j in range(i+1,len(arr)):
    if(arr[j] < arr[i]):
        temp = arr[j]
        arr[j] = arr[i]
        arr[i] = temp

print(arr)

# Time Complexity is O(n^2)
```

```python
arr = [22, 44, 99, 12, 54, 67, 10, 65, 32]

i = 0

while(i<len(arr)):
  minIndex = i
  min = arr[minIndex]

  for j in range(i+1,len(arr)):
    if(arr[j] < min):
      min = arr[j]
      minIndex = j
  for k in range(minIndex,i,-1):
    arr[k] = arr[k-1]
  arr[i] = min
  i=i+1

print(arr)
```
